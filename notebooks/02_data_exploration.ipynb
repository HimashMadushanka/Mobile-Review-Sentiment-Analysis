```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns

data_path = Path("C:/Users/ASUS/Desktop/mobile-review-sentiment-analysis/data/raw/amazon_mobile_reviews.csv")
df = pd.read_csv(data_path)
```

```python
print(df.info())
```

```python
df.isnull().sum()
```

```python
df.dropna(subset=["review_text"], inplace=True)
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df['sentiment'].value_counts()
```

```python
plt.figure(figsize=(7, 4))

sns.countplot(
    data=df,
    x="sentiment",
    order=[
        "negative",
        "neutral",
        "positive"
    ]
)

plt.title("Sentiment Distribution")
plt.xlabel("Sentiment")
plt.ylabel("Number of Reviews")

plt.show()
```

```python
plt.figure(figsize=(7, 4))

sns.countplot(
    data=df,
    x="rating"
)

plt.title("Rating Distribution")

plt.show()
```
