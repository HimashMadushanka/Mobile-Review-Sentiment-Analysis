```python
import sys
from pathlib import Path
import joblib

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT))

from src.data_preprocessing import clean_text

model_path = ROOT / "models" / "traditional_ml" / "best_tfidf_model.joblib"
vectorizer_path = ROOT / "models" / "traditional_ml" / "tfidf_vectorizer.joblib"

for path in (model_path, vectorizer_path):
    if not path.exists():
        raise FileNotFoundError(
            f"Required model artifact not found: {path}. Run notebooks 05 and 06 first."
        )

vectorizer = joblib.load(vectorizer_path)
model = joblib.load(model_path)

examples = [
    "The phone is amazing and the battery lasts all day.",
    "The screen stopped working after two days.",
    "The phone is okay, nothing special.",
    "Don't buy this! Worst customer service ever and poor build quality.",
    "Super fast delivery, works as advertised!"
]

print(f"Loaded model from: {model_path.name}\n")
for text in examples:
    cleaned = clean_text(text)
    features = vectorizer.transform([cleaned])
    prediction = model.predict(features)[0]

    print("Original Review: ", text)
    print("Cleaned Text:    ", cleaned)
    print("Predicted Sentiment:", prediction)
    print("-" * 60)
```
