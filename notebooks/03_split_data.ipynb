```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split

df = pd.read_csv(
    r"D:\\My_Project\\Analyse\\mobile-review-sentiment-analysis\\data\interim\\2_reviews_cleaned.csv"
)

print(df.shape)

df.head()
```

```python
train_df, temp_df = train_test_split(
    df,
    test_size=0.20,
    random_state=42,
    stratify=df["sentiment"]
)

print("Train:", train_df.shape)
print("Temporary:", temp_df.shape)
```

```python
validation_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    random_state=42,
    stratify=temp_df["sentiment"]
)

print("Train:", train_df.shape)
print("Validation:", validation_df.shape)
print("Test:", test_df.shape)
```

```python
train_df.to_csv(
    "D:\\My_Project\\Analyse\\mobile-review-sentiment-analysis\\data\\processed\\train.csv",
    index=False
)

validation_df.to_csv(
    "D:\\My_Project\\Analyse\\mobile-review-sentiment-analysis\\data\\processed\\validation.csv",
    index=False
)

test_df.to_csv(
    "D:\\My_Project\\Analyse\\mobile-review-sentiment-analysis\\data\\processed\\test.csv",
    index=False
)

print("All datasets saved!")
```
